```python
#1. Создайте функцию f(x) = x^6 + 10 * x^3 - 1000
```

```python
def f(x):
    return x**6 + 10 * x**3 - 1000

print(f(2))
```

```python
#2. Создайте список x_points (или массив через numpy), в котором у нас лежат числа от -10 до 10
```

```python
import numpy as np

x_points = np.linspace(-10, 10, 100)

print(x_points)
```

```python
#3. Используйте x_points как аргумент для созданной функции f(x), и выведите на экран результаты функции для каждого значения 
# x_points, также сохраните результаты в y_points (cписок или массив)
```

```python
y_points = f(x_points)

print(y_points)
```

```python
#4. Покажите на графике x_points & y_points
```

```python
import matplotlib.pyplot as plt

plt.figure(figsize=(10, 6))
plt.plot(x_points, y_points, label='f(x)', color='blue')
plt.xlabel('x')
plt.ylabel('f(x)')
plt.title('График функции f(x) = x^6 + 10x^3 - 1000')
plt.axhline(0, color='black', linewidth=0.8)
plt.axvline(0, color='black', linewidth=0.8)
plt.grid(True)
plt.legend()
plt.show()
```

```python
#5. Найдите производную функции f(x) и создайте новую функцию на ее основе как f_diff(x)
```

```python
import sympy as sp

x = sp.symbols('x')

f_expr = x**6 + 10*x**3 - 1000

f_diff_expr = sp.diff(f_expr, x)

print("Производная функции:", f_diff_expr)

f_diff = sp.lambdify(x, f_diff_expr, 'numpy')
```

```python
#6 Чему равна производная при x = 10
```

```python
result = f_diff(10)

print("Производная при x = 10:", result)
```

```python
#7 Покажите на графике f(x) & f_diff(x)
```

```python
plt.figure(figsize=(12, 6))

plt.plot(x_points, f(x_points), label='f(x)', color='blue')
plt.plot(x_points, f_diff(x_points), label="f'(x)", color='red')

plt.xlabel('x')
plt.ylabel('Значение функции')
plt.title('Функция и её производная')
plt.axhline(0, color='black', linewidth=0.8)
plt.axvline(0, color='black', linewidth=0.8)
plt.grid(True)
plt.legend()
plt.show()
```

```python
#8 Найдите уравнение касательной линии в точке x = 9
```

```python
a = 9

y_a = f(a)
slope = f_diff(a)

def tangent(x):
    return y_a + slope * (x - a)

print("f(9) =", y_a)
print("f'(9) =", slope)
print("Уравнение касательной:")
print(f"y = {y_a} + {slope} * (x - 9)")
```

```python
#9 Покажите на графике функцию f(x) & касательную линию в точке x = 9
```

```python
x_tangent = np.linspace(8.5, 9.5, 100)

plt.figure(figsize=(12, 7))

plt.plot(x_points, f(x_points),
         label='f(x)', color='blue')

plt.plot(x_tangent, tangent(x_tangent),
         label='Касательная при x = 9', color='red', linestyle='--')

plt.scatter([9], [f(9)], color='green', s=100,
            label='Точка касания (9, 537731)', zorder=5)

plt.xlabel('x')
plt.ylabel('y')
plt.title('Функция f(x) и касательная в точке x = 9')
plt.grid(True)
plt.legend()
plt.show()
```
