```python
# Импортируем библиотеки

import numpy as np 
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
%matplotlib inline
```

```python
data = pd.read_csv('500_Person_Gender_Height_Weight_Index.csv')
```

```python
# Удаляем ненужный столбец
data.drop('Index', axis = 1, inplace = True)
data.head()
```

```python
data.shape
```

## 1. Создайте колонку BMI (индекс массы тела), которая считается по формуле: BMI = (Вес в кг) / (Рост в метрах)^2. Учтите, что ваш текущий рост в см, не метры.

```python
data['BMI'] = data['Weight'] / ((data['Height'] / 100) ** 2)
display(data.head())
```

## 2. Посчитайте среднее значение, медиану, и моду для роста, веса, BMI.

```python
# Среднее (Mean)
mean_values = data[['Height', 'Weight', 'BMI']].mean()
print("Средние значения:\n", mean_values)
```

```python
# Медиана (Median)
median_values = data[['Height', 'Weight', 'BMI']].median()
print("\nМедианы:\n", median_values)
```

```python
# Мода (Mode)
mode_values = data[['Height', 'Weight', 'BMI']].mode().iloc[0]
print("\nМоды:\n", mode_values)
```

## 3. Найдите range (диапазон, т.е. макс - мин), и стандартное отклонение для роста, веса, BMI. И расскажите что это означает.

```python
for col in ['Height', 'Weight', 'BMI']:
    range_val = data[col].max() - data[col].min()
    std_val = data[col].std()
    print(f"{col}: Range = {range_val:.2f}, Std = {std_val:.2f}")
```

## 4. Найдите 15-й процентиль и 90-й процентиль для роста, веса, BMI. Что это означает простыми словами?

```python
percentiles = data[['Height', 'Weight', 'BMI']].quantile([0.15, 0.90])
print("15-й и 90-й процентили:\n", percentiles)
```

## 5. Создайте гистограмму для роста, веса, BMI.

```python
fig, axes = plt.subplots(1, 3, figsize=(18, 5))
sns.histplot(data['Height'], kde=True, ax=axes[0]).set_title('Распределение роста')
sns.histplot(data['Weight'], kde=True, ax=axes[1]).set_title('Распределение веса')
sns.histplot(data['BMI'], kde=True, ax=axes[2]).set_title('Распределение BMI')
plt.tight_layout()
plt.show()
```

## 6. Создайте box plot для роста, веса, BMI в разрезе пола. Что вы видите, опишите?

```python
fig, axes = plt.subplots(1, 3, figsize=(18, 5))
sns.boxplot(x='Gender', y='Height', data=data, ax=axes[0]).set_title('Рост по полу')
sns.boxplot(x='Gender', y='Weight', data=data, ax=axes[1]).set_title('Вес по полу')
sns.boxplot(x='Gender', y='BMI', data=data, ax=axes[2]).set_title('BMI по полу')
plt.tight_layout()
plt.show()
```

## 7. Посчитайте межквартильный размах для роста, веса, BMI.

```python
for col in ['Height', 'Weight', 'BMI']:
    Q1 = data[col].quantile(0.25)
    Q3 = data[col].quantile(0.75)
    IQR = Q3 - Q1
    print(f"IQR для {col}: {IQR:.2f}")
```

## 8. Есть ли в данных какие-то выбросы для роста, веса, BMI? Как можно их найти? Используя межквартильный размах? Удалите все такие выбросы из таблицы.

```python
# Поиск и удаление выбросов с использованием IQR
def remove_outliers(df, columns):
    df_clean = df.copy()
    for col in columns:
        Q1 = df_clean[col].quantile(0.25)
        Q3 = df_clean[col].quantile(0.75)
        IQR = Q3 - Q1
        lower_bound = Q1 - 1.5 * IQR
        upper_bound = Q3 + 1.5 * IQR
```

```python
# Фильтрация: оставляем только те строки, где значения в пределах границ
        df_clean = df_clean[(df_clean[col] >= lower_bound) & (df_clean[col] <= upper_bound)]
    return df_clean
```

```python
# Удаляем выбросы из таблицы
data_clean = remove_outliers(data, ['Height', 'Weight', 'BMI'])

print(f"Размер данных до удаления выбросов: {data.shape}")
print(f"Размер данных после удаления выбросов: {data_clean.shape}")
```
